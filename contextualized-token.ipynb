```python
from transformers import AutoModel, AutoTokenizer
```

```python
tokenizer = AutoTokenizer.from_pretrained("microsoft/deberta-base")
```

```python
model = AutoModel.from_pretrained("microsoft/deberta-v3-xsmall")
```

```python
tokens = tokenizer('Hello world', return_tensors='pt')
```

```python
output = model(**tokens)[0]
```

```python
output.shape
# (batch_size[number of sentences], sequence_length[number_of_tokens], hidden_size[embedding dimension])
```

```python
# raw output of a language model (for 1 sentence)
output
```

```python
tokens = tokenizer(["Hello world","Hi there"], return_tensors='pt', padding=True)
```

```python
output = model(**tokens)[0]
```

```python
output.shape
```

```python
for token in tokens['input_ids'][0]:
    print(tokenizer.decode(token))
```

```python
# torch.Size([1, 4, 384]) 
# batch_size - 1 (one sentence - hello world)

# sequence_length - 4 (subword tokens) which includes special tokens as well in the count.
# [CLS] Hello world [SEP]

# hidden_size
# model - In microsoft/deberta-v3-xsmall, each token is represented by a 384-dimensional vector
```
